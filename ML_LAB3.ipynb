```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
import numpy as np

def find_s(X, y):
    hypothesis = None
    print("Initial Hypothesis:", hypothesis)
    
    for i in range(len(X)):
        # Process ONLY positive examples ('Yes')
        if y[i] == 'Yes':
            if hypothesis is None:
                hypothesis = X[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != X[i][j]:
                        hypothesis[j] = '?'
                        
        print(f"\nAfter Training Example {i + 1} ({y[i]}):")
        print(hypothesis)
        
    return hypothesis

final_s_hypothesis = find_s(X, y)
print("\nFinal Specific Hypothesis (Find-S):", final_s_hypothesis)
```

Candidate Elimination Algorithm

```python
def candidate_elimination(X, y):
    num_attributes = X.shape[1]
    specific_h = X[0].copy()
    general_h = [["?" for _ in range(num_attributes)] for _ in range(num_attributes)]    
    print("Initial Specific Boundary (S0):", specific_h)
    print("Initial General Boundary (G0):", general_h)   
    for i, h in enumerate(X):
        if y[i] == "Yes":
            for x in range(num_attributes):
                if h[x] != specific_h[x]:
                    specific_h[x] = '?'
                    general_h[x][x] = '?'
        elif y[i] == "No":
            for x in range(num_attributes):
                if h[x] != specific_h[x]:
                    general_h[x][x] = specific_h[x]
                else:
                    general_h[x][x] = '?'
                    
        print(f"\nAfter Training Example {i + 1} ({y[i]}):")
        print("S:", specific_h)
        print("G:", general_h)
    general_h = [g for g in general_h if g != ['?'] * num_attributes]
    return specific_h, general_h

final_s, final_g = candidate_elimination(X, y)

print("\n=== FINAL BOUNDARIES ===")
print("Final Specific Boundary (S):", final_s)
print("Final General Boundary (G):", final_g)

```
